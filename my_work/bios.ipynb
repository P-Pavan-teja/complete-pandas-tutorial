```python
bios_new["first_name"] = bios_new['name'].str.split(" ").str[0]
```

```python

bios_new.to_csv(f'{base_dir}/my_work/saved_csv/bios_new.csv', index=False)
```

```python
bios_new["height_cat"] = bios_new["height_cm"].apply(lambda x: 'missing' if pd.isna(x) else('short' if x < 160.0 else('avrage' if x < 200 else 'tall')))
```

```python
bios_new.head(2)
```

```python
def cat_hight_weight(row):
    if pd.isna(row['height_cm']) | pd.isna(row['weight_kg']):
        return 'missing values'
    elif row['height_cm'] < 175.0 and row['weight_kg'] < 70.0:
        return 'Light weight'
    elif row['height_cm'] < 200.0 or row['weight_kg'] < 95.0:
        return 'medium weight'
    else:
        return 'Heavy weight'
```

```python
bios_new["height_cat_weigh"] = bios_new.apply(cat_hight_weight, axis=1)
bios_new.head(2)
```
